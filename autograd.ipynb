*To prevent tracking gradient*

```python
x1 = torch.tensor([1,5,8,9,16],dtype=torch.float32,requires_grad=True)
print(x1)
```

```python
x1.requires_grad_(False)
print(x1)
```

```python
# detach method
x2 = torch.tensor([7,8,9,5],dtype=torch.float32,requires_grad=True)
print(x2)
y = x2.detach()
print(y)
```

```python
with torch.no_grad():
    y = x2+3
    print(y)
```

```python
y = x2+1
print(y)
```
