```python
import torch
```

```python
# Input
x = torch.tensor(2.0)
```

```python
# Weight
w = torch.tensor(3.0, requires_grad=True)
```

```python
# Target
target = torch.tensor(10.0)
```

```python
learning_rate = 0.1
```

```python
for epoch in range(10):

    # 1. Forward pass
    prediction = w * x

    # 2. Calculate loss
    loss = (prediction - target) ** 2

    # 3. Backpropagation
    loss.backward()

    # 4. Update weight manually
    with torch.no_grad():
        w -= learning_rate * w.grad

    # 5. Clear gradient
    w.grad.zero_()

    print(f"Epoch: {epoch + 1}, ")
    print(f"Weight: {w.item():.4f}, ")
    print(f"Loss: {loss.item():.4f}")
    
```
