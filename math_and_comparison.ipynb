*Matrix Multiplication*

```python
x1 = torch.rand((2,5))
print(x1)
x2 = torch.rand((5,3))
print(x2)
```

```python
x3 =torch.mm(x1,x2)
print(x3)
```

*Matrix Exponenziation*

```python
y1 = torch.randint(0,5,(5,5))
print(y1)
c = y1.matrix_power(3)
print(c)
```

*Batch Matrix Multiplication*

```python
batch = 32
n = 10
m = 20
p = 30

t1 = torch.rand((batch,n,m))
t2 = torch.rand((batch,m,p))
out_bmm = torch.bmm(t1,t2)
print(out_bmm)
```
