```python
import torch
```

```python
x = torch.arange(9)
x1 = x.view(3,3)
print(x1)
x2 = x.reshape(3,3)
print(x2)
```

```python
y = x2.t()
print(y)
```

```python
print(y.contiguous().view(9))
```

```python
x1 = torch.rand(2,5)
x2 = torch.rand(2,5)
print(x1)
print(x2)

print(torch.cat((x1,x2),dim=0).shape)
print(torch.cat((x1,x2)))
print(torch.cat((x1,x2),dim=1).shape)
```

```python
z = x1.view(-1)
print(z)
print(z.shape)
```

```python
batch = 64
c = torch.rand((batch,2,5))
print(c)
```

```python
d = c.view(batch,-1)
print(d)
print(d.shape)
```

```python
d = c.permute(0,2,1)
print(d)
print(d.shape)
```

```python
e = torch.arange(10)
print(e.unsqueeze(0))
print(e.unsqueeze(0).shape)
print(e.unsqueeze(1))
print(e.unsqueeze(1).shape)
```

```python
f = torch.arange(10).unsqueeze(0).unsqueeze(1)
print(f)
```

```python
g = f.squeeze(1)
print(g)
print(g.shape)
```
